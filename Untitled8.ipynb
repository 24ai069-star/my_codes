```python
import re
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense, Embedding
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

data = {
    "English": [
        "how are you", "where are you going", "this is beautiful", "call me as soon as possible",
        "please help me", "I am fine", "what are you doing", "come here quickly",
        "do you understand me", "good morning my friend", "where is your home", "i love music",
        "let us go now", "speak in english please", "it is very hot today"
    ],
    "Hinglish": [
        "aap kaise ho", "aap kaha ja rahe ho", "yeh bahut sundar hai", "mujhe jaldi se jaldi call karo",
        "kripya meri madad karein", "main theek hoon", "aap kya kar rahe ho", "yahan jaldi aao",
        "kya aap mujhe samajhte ho", "shubh prabhat mere dost", "aapka ghar kaha hai", "mujhe music pasand hai",
        "chalo ab chalte hain", "kripya english mein bolo", "aaj bahut garmi hai"
    ]
}
df = pd.DataFrame(data)

def preprocess_sentence(text):
    if not isinstance(text, str):
        return ""
    text = text.lower().strip()
    text = re.sub(r"([?.!,¿])", r" \1 ", text)
    text = re.sub(r'[" ]+', " ", text)
    text = re.sub(r"[^a-zA-Z?.!,¿]+", " ", text)
    text = text.strip()
    return '<start> ' + text + ' <end>'

df['english_clean'] = df['English'].apply(preprocess_sentence)
df['hinglish_clean'] = df['Hinglish'].apply(preprocess_sentence)

def tokenize_and_pad(texts, max_len=None):
    tokenizer = Tokenizer(filters='!"#$%&()*+,-./:;=?@[\\]^_`{|}~')
    tokenizer.fit_on_texts(texts)
    sequences = tokenizer.texts_to_sequences(texts)
    padded_sequences = pad_sequences(sequences, padding='post', maxlen=max_len)
    return padded_sequences, tokenizer

input_data, input_tokenizer = tokenize_and_pad(df['english_clean'].tolist())
target_data, target_tokenizer = tokenize_and_pad(df['hinglish_clean'].tolist())

num_input_tokens = len(input_tokenizer.word_index) + 1
num_target_tokens = len(target_tokenizer.word_index) + 1
max_encoder_seq_length = input_data.shape[1]
max_decoder_seq_length = target_data.shape[1]

encoder_inputs = Input(shape=(None,))
x = Embedding(num_input_tokens, 256, mask_zero=True)(encoder_inputs)
encoder_outputs, state_h, state_c = LSTM(256, return_state=True)(x)
encoder_states = [state_h, state_c]

decoder_inputs = Input(shape=(None,))
x = Embedding(num_target_tokens, 256, mask_zero=True)(decoder_inputs)
decoder_lstm = LSTM(256, return_sequences=True, return_state=True)
decoder_outputs, _, _ = decoder_lstm(x, initial_state=encoder_states)
decoder_dense = Dense(num_target_tokens, activation='softmax')
decoder_outputs = decoder_dense(decoder_outputs)

model = Model([encoder_inputs, decoder_inputs], decoder_outputs)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')

decoder_input_data = target_data[:, :-1]
decoder_target_data = np.expand_dims(target_data[:, 1:], -1)

model.fit([input_data, decoder_input_data], decoder_target_data, batch_size=64, epochs=30, validation_split=0.1)

encoder_model = Model(encoder_inputs, encoder_states)

decoder_state_input_h = Input(shape=(256,))
decoder_state_input_c = Input(shape=(256,))
decoder_states_inputs = [decoder_state_input_h, decoder_state_input_c]

dec_emb_out = Embedding(num_target_tokens, 256, mask_zero=True)(decoder_inputs)
decoder_outputs, state_h_dec, state_c_dec = decoder_lstm(dec_emb_out, initial_state=decoder_states_inputs)
decoder_states_outputs = [state_h_dec, state_c_dec]
decoder_outputs = decoder_dense(decoder_outputs)

decoder_model = Model([decoder_inputs] + decoder_states_inputs, [decoder_outputs] + decoder_states_outputs)

def decode_sequence(input_seq):
    states_value = encoder_model.predict(input_seq, verbose=0)
    target_seq = np.zeros((1, 1))
    target_seq[0, 0] = target_tokenizer.word_index['<start>']
    stop_condition = False
    decoded_sentence = ''
    while not stop_condition:
        output_tokens, h, c = decoder_model.predict([target_seq] + states_value, verbose=0)
        sampled_token_index = np.argmax(output_tokens[0, -1, :])
        sampled_char = target_tokenizer.index_word.get(sampled_token_index, '')
        if sampled_char == '<end>' or len(decoded_sentence.split()) > max_decoder_seq_length:
            stop_condition = True
        else:
            decoded_sentence += ' ' + sampled_char
        target_seq = np.zeros((1, 1))
        target_seq[0, 0] = sampled_token_index
        states_value = [h, c]
    return decoded_sentence.strip()

test_sentences = [
    "how are you today",
    "where are you going",
    "this is extremely beautiful",
    "call me as soon as possible",
    "please help me with this task"
]

for sent in test_sentences:
    cleaned_sent = preprocess_sentence(sent)
    seq = input_tokenizer.texts_to_sequences([cleaned_sent])
    padded = pad_sequences(seq, maxlen=max_encoder_seq_length, padding='post')
    translation = decode_sequence(padded)
    print(f"English: {sent}")
    print(f"Hinglish: {translation}\n")
```
